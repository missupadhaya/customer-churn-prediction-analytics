# Customer Churn Prediction — Exploratory Data Analysis

This notebook explores the included **synthetic** dataset. It is intended for learning and portfolio demonstration.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/customer_churn.csv')
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
df.isna().sum()
df['churn'].value_counts(normalize=True).rename('share')
```

```python
sns.countplot(data=df, x='churn')
plt.title('Churn distribution (0=retained, 1=churned)')
plt.show()
```

```python
sns.boxplot(data=df, x='churn', y='monthly_charges')
plt.title('Monthly charges by churn')
plt.show()
```

```python
sns.barplot(data=df, x='contract_type', y='churn')
plt.title('Churn share by contract type')
plt.xticks(rotation=20)
plt.show()
```

```python
sns.heatmap(df[['tenure_months','monthly_charges','support_tickets','service_outages','churn']].corr(), annot=True, cmap='coolwarm')
plt.title('Numeric feature correlation')
plt.show()
```
